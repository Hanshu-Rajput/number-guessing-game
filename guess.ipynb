```python
# Game begin
import random
n=random.randint(1,100)
a=-1
guesses=1
while(a!=n):
   
    a=int(input("Enter the number you guess :"))
    if(a>n):
        print("Enter a lower number, please!")
        guesses+=1
    elif(a<n):
        print("Enter a higher number, please!")
        guesses+=1
print(f"You have guessed the number {n} correctly in {guesses} attempts .....")
```
